## 1.1 Number System and Basic Arithmetic Operations

### Example 1.1.1: Classifying numbers

```python
from fractions import Fraction

numbers = {"-3": -3, "0": 0, "7": 7, "2/5": Fraction(2, 5),
           "sqrt(2)": 2 ** 0.5, "pi": 3.141592653589793, "0.333...": Fraction(1, 3)}
irrational = {"sqrt(2)", "pi"}        # never end, never repeat

for name, x in numbers.items():
    kinds = []
    if name in irrational:
        kinds.append("irrational")
    else:
        if x == int(x) and x >= 1: kinds.append("natural")
        if x == int(x) and x >= 0: kinds.append("whole")
        if x == int(x):            kinds.append("integer")
        kinds.append("rational")
    kinds.append("real")
    print(f"{name:>9}: {', '.join(kinds)}")
```

### Example 1.1.8: HCF and LCM

```python
import math

def prime_factors(n):
    factors, d = [], 2
    while n > 1:
        while n % d == 0:
            factors.append(d)
            n //= d
        d += 1
    return factors

print("12 =", prime_factors(12))
print("18 =", prime_factors(18))
hcf, lcm = math.gcd(12, 18), math.lcm(12, 18)
print("HCF =", hcf, " LCM =", lcm)
print("Check HCF x LCM = 12 x 18:", hcf * lcm == 12 * 18)
```

### Example 1.1.12: Terminating or repeating?

```python
from fractions import Fraction

def terminates(fr):
    d = fr.denominator          # Fraction is already in lowest terms
    for p in (2, 5):
        while d % p == 0:
            d //= p
    return d == 1

for fr in [Fraction(7, 8), Fraction(2, 3)]:
    kind = "terminating" if terminates(fr) else "repeating"
    print(f"{fr} = {float(fr)} ... {kind}")
```

## 1.3 Rates, Ratios, Proportions and Percentages

### Example 1.3.1: Simplifying ratios

```python
import math

def simplify_ratio(a, b):
    h = math.gcd(a, b)
    return a // h, b // h

print("(a) 24:36 ->", simplify_ratio(24, 36))
print("(b) 50 cm : 2 m = 50:200 ->", simplify_ratio(50, 2 * 100))
print("(c) 1/2 : 1/3, times 6 ->", (6 // 2, 6 // 3))
```

## 1.4 Types and Sources of Data

### Example 1.4.1: Classifying variables

```python
variables = {
    "gender":        ("words", None),
    "brothers":      ("number", "counted"),
    "height":        ("number", "measured"),
    "blood group":   ("words", None),
    "temperature":   ("number", "measured"),
    "cars in a lot": ("number", "counted"),
}
for name, (answer, how) in variables.items():
    if answer == "words":
        kind = "qualitative"
    elif how == "counted":
        kind = "quantitative, discrete"
    else:
        kind = "quantitative, continuous"
    print(f"{name:<14} -> {kind}")
```

### Example 1.4.2: Trickier cases

```python
# Discrete: only separate values are possible. Continuous: any value in a range.
in_between_possible = {"shoe size": False, "time to solve puzzle": True,
                       "goals in a match": False, "monthly rainfall": True}
for name, between in in_between_possible.items():
    print(f"{name:<22} -> {'continuous' if between else 'discrete'}")
```

### Example 1.4.3: Primary or secondary?

```python
studies = [
    ("student asks 50 classmates", "self"),
    ("researcher uses census figures", "Pakistan Bureau of Statistics"),
]
for what, collector in studies:
    kind = "primary" if collector == "self" else "secondary"
    print(f"{what:<32} -> {kind} (collected by {collector})")
```

### Example 1.4.4: Judging a source

```python
sources = {
    "State Bank of Pakistan website": {"who": True, "how": True, "when": True},
    "social-media post":              {"who": False, "how": False, "when": False},
}
for name, checks in sources.items():
    score = sum(checks.values())
    print(f"{name:<31} passes {score}/3 checks")
print("More reliable:", max(sources, key=lambda s: sum(sources[s].values())))
```

```python
survey = {
    "gender": "qualitative (nominal)",
    "subjects studied daily": "quantitative, discrete",
    "hours of study per day": "quantitative, continuous",
}
print("Method: questionnaire -> primary data")
for var, kind in survey.items():
    print(f"  {var:<24} {kind}")
```

## 1.6 Tabular and Graphical Presentation of Data

### Example 1.6.1: Making a grouped frequency table

```python
marks = [12, 25, 37, 18, 42, 29, 33, 15, 27, 39,
         21, 45, 31, 24, 36, 19, 28, 41, 34, 26]
classes = [(10, 19), (20, 29), (30, 39), (40, 49)]
n = len(marks)
cum = 0
print(f"{'Class':<8}{'Bound.':<12}{'Mid':>5}{'f':>4}{'Rel f':>7}{'Cum f':>7}")
for lo, hi in classes:
    f = sum(1 for m in marks if lo <= m <= hi)
    cum += f
    cls = f"{lo}-{hi}"
    bound = f"{lo - 0.5}-{hi + 0.5}"
    print(f"{cls:<8}{bound:<12}{(lo + hi) / 2:>5}{f:>4}{f / n:>7.2f}{cum:>7}")
print("Total:", n)
```

### Example 1.6.2: Frequency table for discrete data

```python
goals = [0, 1, 2, 1, 3, 0, 2, 1, 1, 4, 2, 0, 1, 3, 2, 1, 0, 2, 1, 1]
freq = {}
for g in goals:
    freq[g] = freq.get(g, 0) + 1
for g in sorted(freq):
    print(f"Goals {g}: {freq[g]}")
print("Total matches:", sum(freq.values()))
print("Most common:", max(freq, key=freq.get), "goal")
```

### Example 1.6.3: Drawing a histogram and frequency polygon

```python
import matplotlib.pyplot as plt
marks = [12, 25, 37, 18, 42, 29, 33, 15, 27, 39,
         21, 45, 31, 24, 36, 19, 28, 41, 34, 26]
bounds = [9.5, 19.5, 29.5, 39.5, 49.5]
plt.figure(figsize=(6, 3.8))
counts, _, _ = plt.hist(marks, bins=bounds, color="lightseagreen", edgecolor="black")
mids = [14.5, 24.5, 34.5, 44.5]
plt.plot(mids, counts, "o-", color="darkred", label="frequency polygon")
plt.xticks(bounds)
plt.xlabel("Marks (class boundaries)"); plt.ylabel("Frequency")
plt.title("Histogram of marks"); plt.legend()
plt.show()
```

### Example 1.6.4: Ogive (cumulative frequency curve)

```python
import matplotlib.pyplot as plt
upper = [9.5, 19.5, 29.5, 39.5, 49.5]
cum_f = [0, 4, 11, 17, 20]
plt.figure(figsize=(6, 3.8))
plt.plot(upper, cum_f, "o-", color="teal")
plt.plot([29.5, 29.5, 5], [0, 11, 11], "r--")
plt.xticks(upper); plt.grid(alpha=0.3)
plt.xlabel("Marks (upper boundary)"); plt.ylabel("Cumulative frequency")
plt.title("Ogive")
plt.show()
print("Below 29.5:", cum_f[2], " Above 39.5:", cum_f[-1] - cum_f[3])
```

### Example 1.6.9: Choosing the right graph

```python
def choose_graph(purpose):
    rules = {"parts of a whole": "pie chart", "change over time": "line graph",
             "continuous data in classes": "histogram", "separate categories": "bar chart"}
    return rules[purpose]

tasks = {"market share of networks": "parts of a whole",
         "monthly temperature": "change over time",
         "heights of 100 students": "continuous data in classes",
         "students per department": "separate categories"}
for task, purpose in tasks.items():
    print(f"{task:<26} -> {choose_graph(purpose)}")
```

### Example 1.7.10: The average-speed trap

```python
d = 60
t_go, t_back = d / 30, d / 60
average = (2 * d) / (t_go + t_back)
print("Times:", t_go, "h and", t_back, "h")
print("Average speed =", average, "km/h  (not", (30 + 60) / 2, ")")
```
