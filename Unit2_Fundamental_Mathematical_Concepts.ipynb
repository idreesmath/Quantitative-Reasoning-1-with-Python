### Example 2.1.8: Is it a right triangle?

```python
def triangle_type(a, b, c):
    a, b, c = sorted([a, b, c])          # c is now the longest side
    if a**2 + b**2 == c**2:
        return "right-angled"
    return "obtuse" if c**2 > a**2 + b**2 else "acute"

for sides in [(7, 24, 25), (5, 6, 8)]:
    print(sides, "->", triangle_type(*sides))
```

## 2.2 Sets and their Operations

### Example 2.2.1: Roster and set-builder forms

```python
import string
vowels = {ch for ch in string.ascii_lowercase if ch in "aeiou"}
print("(a)", sorted(vowels))
squares = {x**2 for x in range(1, 6)}
print("(b) {x^2 : x = 1..5} =", sorted(squares))
print("(c)", {x for x in range(1, 6)})
```

### Example 2.2.2: Types of sets

```python
a = {x for x in range(1, 10) if x < 1}      # natural numbers less than 1
print("(a)", a, " empty?", len(a) == 0)
b = {0}
print("(b)", b, " elements:", len(b))
A, B = {1, 2, 3}, {"a", "b", "c"}
print("(c) equivalent?", len(A) == len(B), " equal?", A == B)
```

### Example 2.2.3: Subsets and power sets

```python
from itertools import combinations

def power_set(s):
    s = sorted(s)
    return [set(c) for r in range(len(s) + 1) for c in combinations(s, r)]

for S in [{"a", "b"}, {1, 2, 3}]:
    P = power_set(S)
    print(f"Subsets of {sorted(S)} ({len(P)} = 2^{len(S)}):")
    print("  ", [sorted(p) for p in P])
```

### Example 2.2.4: Set operations

```python
U = set(range(1, 11))
A = {1, 2, 3, 4, 5}
B = {2, 4, 6, 8, 10}
print("A u B =", A | B)
print("A n B =", A & B)
print("A - B =", A - B, "  B - A =", sorted(B - A))   # sorted() lists in order
print("A'    =", U - A)
print("B'    =", U - B)
```

### Example 2.2.5: Difference as intersection with a complement

```python
U = set(range(1, 11))
A, B = {1, 2, 3, 4, 5}, {2, 4, 6, 8, 10}
print("A - B  =", A - B)
print("A n B' =", A & (U - B))
print("Equal?", A - B == A & (U - B))
```

### Example 2.2.6: Verifying De Morgan's law

```python
U = set(range(1, 11))
A, B = {1, 2, 3, 4, 5}, {2, 4, 6, 8, 10}
lhs = U - (A | B)
rhs = (U - A) & (U - B)
print("(A u B)' =", sorted(lhs))
print("A' n B'  =", sorted(rhs))
print("De Morgan holds:", lhs == rhs)
```

### Example 2.2.7: Survey problem with a Venn diagram

```python
total, tea, coffee, both = 60, 35, 30, 12
only_tea = tea - both
only_coffee = coffee - both
at_least_one = tea + coffee - both
neither = total - at_least_one
print("Only tea:", only_tea, " Only coffee:", only_coffee)
print("At least one:", at_least_one, " Neither:", neither)
print("Check:", only_tea + both + only_coffee + neither)
```

### Example 2.2.8: Three sets

```python
total = 100
M, P, C = 45, 40, 35
MP, MC, PC, MPC = 15, 10, 12, 5
union = M + P + C - MP - MC - PC + MPC
print("At least one subject:", union)
print("None               :", total - union)
print("Only Maths         :", M - MP - MC + MPC)
```

### Example 2.3.1: Cartesian product

### Example 2.3.2: A relation given by a rule

### Example 2.3.3: Function or not?

```python
def is_function(relation):
    inputs = [a for a, b in relation]
    return len(inputs) == len(set(inputs))    # no input repeated

R1 = {(1, 2), (2, 3), (3, 4)}
R2 = {(1, 2), (1, 3), (2, 4)}
print("R1 is a function:", is_function(R1))
print("R2 is a function:", is_function(R2))
```

### Example 2.3.4: Vertical line test

```python
from sympy import symbols, solve
x, y = symbols("x y")
for name, curve in [("y = x^2", y - x**2), ("x^2 + y^2 = 4", x**2 + y**2 - 4)]:
    ys = solve(curve.subs(x, 0), y)          # vertical line x = 0
    verdict = "function" if len(ys) <= 1 else "NOT a function"
    print(f"{name:<14} meets x = 0 at y = {ys} -> {verdict}")
```

### Example 2.3.7: Slope and equation of a line

```python
from sympy import symbols, Rational, expand
x1, y1, x2, y2 = 1, 3, 3, 7
m = Rational(y2 - y1, x2 - x1)
x = symbols("x")
line = expand(y1 + m * (x - x1))
print("Slope m =", m)
print("Equation: y =", line)
print("Check at x = 3:", line.subs(x, 3))
```

### Example 2.4.7: Factorising trinomials

```python
from sympy import symbols, factor
x = symbols("x")

def find_pair(product, total):
    for p in range(-abs(product), abs(product) + 1):
        if p != 0 and product % p == 0 and p + product // p == total:
            return p, product // p

print("x^2+5x+6 : numbers", find_pair(6, 5), "->", factor(x**2 + 5*x + 6))
print("x^2-x-12 : numbers", find_pair(-12, -1), "->", factor(x**2 - x - 12))
```

### Example 2.5.10: Quadratic formula (rational roots)

### Example 2.5.11: Quadratic formula (irrational roots)

```python
from sympy import symbols, solve, N
x = symbols("x")
roots = solve(x**2 + 2*x - 4, x)
print("Discriminant:", 2**2 - 4*1*(-4))
print("Exact roots :", roots)
print("Decimals    :", [round(float(N(r)), 3) for r in roots])
```

### Example 2.5.12: Nature of roots

```python
def nature(a, b, c):
    D = b**2 - 4*a*c
    if D > 0:  return D, "two real roots"
    if D == 0: return D, "one repeated root"
    return D, "no real roots"

print("x^2 - 6x + 9 :", nature(1, -6, 9))
print("x^2 + x + 2  :", nature(1, 1, 2))
```

### Example 2.6.1: Comparing mobile plans
