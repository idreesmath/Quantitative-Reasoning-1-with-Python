## 0.6 Repetition: Loops and Comprehensions

### Example 0.6.1: Looping over a list

```python
marks = [45, 72, 88, 39]
for m in marks:
    if m >= 50:
        print(m, "Pass")
    else:
        print(m, "Fail")
```

### Example 0.6.2: Using range

```python
for i in range(1, 11):
    print(f"7 x {i:>2} = {7 * i:>3}")
```

### Example 0.6.3: Accumulating a sum

```python
total = 0
for i in range(1, 101):
    total += i              # same as total = total + i
print("Sum by loop    :", total)
n = 100
print("Sum by formula :", n * (n + 1) // 2)
```

### Example 0.6.4: A while loop

```python
amount = 10000
years = 0
while amount < 20000:
    amount = amount * 1.10
    years += 1
print("Years needed:", years)
print(f"Amount then : Rs. {amount:,.2f}")
```

### Example 0.6.5: List comprehensions

```python
squares = [x**2 for x in range(1, 11)]
print("Squares:", squares)
even_squares = [s for s in squares if s % 2 == 0]
print("Even squares:", even_squares)
```

### Example 0.6.6: Counting frequencies

```python
data = [2, 3, 1, 4, 2, 3, 2, 5]
freq = {}
for x in data:
    freq[x] = freq.get(x, 0) + 1
for value in sorted(freq):
    print(value, "occurs", freq[value], "times")
```

## 0.7 Functions

### Example 0.7.1: Defining a function

```python
import math

def circle_area(r):
    return math.pi * r**2

print(f"r = 7: area = {circle_area(7):.2f}")
print(f"r = 3: area = {circle_area(3):.2f}")
```

### Example 0.7.2: Default values

```python
def simple_interest(P, R, T=1):
    return P * R * T / 100

print(simple_interest(50000, 8))       # T = 1 (default)
print(simple_interest(50000, 8, 3))    # T = 3 years
```

### Example 0.7.3: Returning several values

```python
def summary(values):
    mean = sum(values) / len(values)
    return mean, min(values), max(values)

m, low, high = summary([3, 5, 7, 7, 8])
print("Mean:", m, " Smallest:", low, " Largest:", high)
```

### Example 0.7.4: Lambda functions

```python
f = lambda x: 3*x - 2
print("f(0) =", f(0))
print("f(2) =", f(2))
print("f(-1) =", f(-1))
print([f(x) for x in range(5)])     # f applied to 0, 1, 2, 3, 4
```

### Example 0.7.5: Functions working together

```python
def percent_change(old, new):
    return (new - old) / old * 100

print(f"400 -> 500  : {percent_change(400, 500):.1f}%")
print(f"2500 -> 2000: {percent_change(2500, 2000):.1f}%")
```

## 0.9 Algebra with SymPy

### Example 0.9.1: Expanding, factorising and simplifying

```python
from sympy import symbols, expand, factor, simplify
x = symbols("x")
print(expand((x + 3)*(x - 5)))
print(factor(x**2 - 25))
print(factor(2*x**2 + 7*x + 3))
print(simplify((x**2 - 9)/(x + 3)))
```

### Example 0.9.2: Solving equations

```python
from sympy import symbols, Eq, solve
x, y = symbols("x y")
print(solve(Eq(3*x + 5, 20), x))
print(solve(Eq(x**2 - 5*x + 6, 0), x))
print(solve([Eq(x + y, 10), Eq(x - y, 2)], [x, y]))
```

### Example 0.9.3: Exact values and substitution

```python
from sympy import symbols, solve, N
x = symbols("x")
roots = solve(x**2 + 2*x - 4, x)
print("Exact  :", roots)
print("Decimal:", [round(float(N(r)), 4) for r in roots])
expr = x**2 - 3*x + 2
print("Value at x = -2:", expr.subs(x, -2))
```

### Example 0.9.4: Solving inequalities

```python
from sympy import symbols, solve_univariate_inequality
x = symbols("x", real=True)
print(solve_univariate_inequality(2*x - 3 < 7, x))
print(solve_univariate_inequality(x**2 - 9 < 0, x))
```

### Example 0.10.1: NumPy arrays

### Example 0.10.2: A line graph

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-3, 3, 100)
y = 2*x + 1
plt.figure(figsize=(5, 3.5))
plt.plot(x, y, label="y = 2x + 1")
plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(0, color="black", linewidth=0.8)
plt.title("Graph of y = 2x + 1")
plt.xlabel("x"); plt.ylabel("y")
plt.grid(True); plt.legend()
plt.show()
```

### Example 0.10.3: Bar chart and pie chart

```python
import matplotlib.pyplot as plt
sports = ["Cricket", "Football", "Hockey", "Others"]
students = [50, 30, 20, 20]
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(9, 3.5))
ax1.bar(sports, students, color="steelblue")
ax1.set_title("Bar chart"); ax1.set_ylabel("Students")
ax2.pie(students, labels=sports, autopct="%1.0f%%")
ax2.set_title("Pie chart")
plt.show()
```

### Example 0.10.4: Histogram

```python
import matplotlib.pyplot as plt
marks = [12, 25, 37, 18, 42, 29, 33, 15, 27, 39,
         21, 45, 31, 24, 36, 19, 28, 41, 34, 26]
plt.figure(figsize=(5, 3.5))
plt.hist(marks, bins=[9.5, 19.5, 29.5, 39.5, 49.5],
         color="seagreen", edgecolor="black")
plt.title("Marks of 20 students")
plt.xlabel("Marks"); plt.ylabel("Frequency")
plt.show()
```

### Example 0.10.5: Probability with SciPy

```python
from scipy import stats
p2 = stats.binom.pmf(2, 4, 0.5)        # P(X = 2), n = 4, p = 0.5
below = stats.norm.cdf(76, 60, 8)      # P(X < 76), mu = 60, sigma = 8
print(f"P(exactly 2 heads in 4 tosses) = {p2:.4f}")
print(f"P(X < 76) = {below:.4f}")
```
